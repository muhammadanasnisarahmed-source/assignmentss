### **Question 32**
Director 'Christopher Nolan' ki un movies ke naam aur release year nikalein jinka 'Runtime' 140 minutes se zyada hai.

### **Question 33**
Aise actors (Star1) ki list nikalein jinhone 'Action' ya 'Sci-Fi' genre mein kaam kiya hai AUR unki movies ki 'IMDB_Rating' kam az kam 8.0 hai.

```python
genre_text = df["Genre"].fillna("").str.split(",")
has_action_or_scifi = genre_text.apply(
    lambda genres: any(g.strip().lower() in {"action", "sci-fi"} for g in genres)
)
mask = has_action_or_scifi & (df["IMDB_Rating"] >= 8.0)
display(df.loc[mask, ["Star1", "Series_Title", "Genre", "IMDB_Rating"]]
          .drop_duplicates(subset=["Star1"])
          .sort_values("Star1"))
```

### **Question 35**

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# Compare individual Drama and Action genre memberships.
genre_text = df["Genre"].fillna("")
drama_gross = df.loc[genre_text.str.contains(r"(^|,\s*)Drama(\s*,|$)", regex=True), "Gross"].dropna()
action_gross = df.loc[genre_text.str.contains(r"(^|,\s*)Action(\s*,|$)", regex=True), "Gross"].dropna()
plt.figure(figsize=(8, 5))
plt.boxplot([drama_gross, action_gross], tick_labels=["Drama", "Action"], showfliers=True)
plt.title("Gross Revenue: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.ticklabel_format(style="plain", axis="y")
plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.nlargest(5, "IMDB_Rating").copy()
top5["IMDB_Rating_Scaled"] = top5["IMDB_Rating"] * 10
x = range(len(top5))
width = 0.38
plt.figure(figsize=(11, 6))
plt.bar([i - width/2 for i in x], top5["Meta_score"], width=width, label="Meta Score")
plt.bar([i + width/2 for i in x], top5["IMDB_Rating_Scaled"], width=width, label="IMDb Rating (scaled to 100)")
plt.xticks(list(x), top5["Series_Title"], rotation=30, ha="right")
plt.title("Top 5 Highest-Rated Movies: Meta Score vs IMDb Rating")
plt.ylabel("Score out of 100")
plt.legend()
plt.tight_layout()
plt.show()
```
